```python
import numpy as np
import matplotlib.pyplot as plt

#the table is from the one written at the whiteboard around 2 weeks ago
x1 = np.array([1, 2, 1, 2], dtype=float)
x2 = np.array([1, 1, 2, 2], dtype=float)
y  = np.array([-1, 5, 8, 14], dtype=float)

X = np.column_stack([x1, x2, np.ones_like(x1)])

w = np.zeros(3)
alpha = 0.01
tolerance = 1e-5

loss_hist = []
steepness_hist = []

while True:
    err = X @ w - y
    grad = 2 * X.T @ err
    steepness = np.linalg.norm(grad)

    loss_hist.append(err @ err)
    steepness_hist.append(steepness)

    if steepness <= tolerance:
        break

    w -= alpha * grad

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(loss_hist, color='crimson')
axes[0].set_yscale('log')
axes[0].set_title("Loss E over Iterations")
axes[0].set_xlabel("Iteration")
axes[0].set_ylabel("Error E")
axes[0].grid(True, which="both", ls="--")

axes[1].plot(steepness_hist, color='teal')
axes[1].axhline(y=tolerance, color='black', linestyle=':', label=r'Target Steepness $= 10^{-5}$')
axes[1].set_yscale('log')
axes[1].set_title("Gradient Steepness over Iterations")
axes[1].set_xlabel("Iteration")
axes[1].set_ylabel("Gradient Steepness")
axes[1].legend()
axes[1].grid(True, which="both", ls="--")

plt.tight_layout()
plt.show()

print(f"Converged in {len(loss_hist)} iterations: b1 = {w[0]:.6f}, b2 = {w[1]:.6f}, c = {w[2]:.6f}")
print(f"Final Gradient Steepness: {steepness:.6e}")
```
